## Data Normalization

Min-Max normalization was applied to the effectiveness-related numerical variables so that all values are converted to a common scale between 0 and 1.

This is important because the variables originally had different ranges, such as quiz scores, feedback scores, watch time, and percentage-based measures.

The dropout rate was converted into not_drop_rate using:

not_drop_rate = 1 - dropout_rate

This makes the direction of all effectiveness metrics consistent, where higher values represent better outcomes.

The normalized variables will now be used to calculate the overall Instructor Effectiveness Score.

```python
df.columns
```

```python
from sklearn.preprocessing import MinMaxScaler

p_cols=['completion_rate',
       'avg_score_improvement', 'avg_quiz_score','avg_watch_time', 'assignment_submission_rate', 'forum_activity_rate',
       'avg_feedback_score', 'feedback_response_rate']

scaler=MinMaxScaler()

df[p_cols]=scaler.fit_transform(df[p_cols])


df['dropout_rate']=scaler.fit_transform(df[['dropout_rate']])
df['not_drop_rate']=1-df['dropout_rate']

df.drop(columns=['dropout_rate'], inplace=True)
```

```python
df
```

## Instructor Effectiveness Score

The normalized variables were divided into three dimensions, and the mean of the relevant variables was used to calculate each component score.

- **Outcome Score**: Mean of completion_rate, avg_score_improvement, avg_quiz_score, and not_drop_rate.
- **Engagement Score**: Mean of avg_watch_time, assignment_submission_rate, and forum_activity_rate.
- **Feedback Score**: Mean of avg_feedback_score and feedback_response_rate.

The three component scores were then combined using weighted averaging:

- Outcome Score: **50%**
- Engagement Score: **30%**
- Feedback Score: **20%**

The final effectiveness score was calculated as:

**Final Effectiveness = (Outcome Score X 0.50) + (Engagement Score X 0.30) + (Feedback Score X 0.20)**

The final score ranges from 0 to 1, where a higher value indicates better overall effectiveness.

```python
df["outcome_score"]=df[['completion_rate',
       'avg_score_improvement', 'avg_quiz_score','not_drop_rate']].mean(axis=1)
```

## Instructor-Level Dataset

The original dataset contains multiple rows for different course batches, and an instructor can be associated with multiple batches. Therefore, the analysis needs to be moved from the batch level to the instructor level.

First, an `instructor_df` was created by grouping the batch-level data by `instructor_id` and calculating the mean `final_effectiveness` for each instructor.

A separate `instructor_features_df` DataFrame was then created by grouping the original dataset by `instructor_id` and calculating the mean of the relevant numerical features across all batches handled by each instructor.

The two DataFrames were then merged using `instructor_id` to create the complete instructor-level dataset.

The final instructor-level dataset contains:

- `instructor_id`
- Mean `completion_rate`
- Mean `avg_score_improvement`
- Mean `avg_quiz_score`
- Mean `not_drop_rate`
- Mean `avg_watch_time`
- Mean `assignment_submission_rate`
- Mean `forum_activity_rate`
- Mean `avg_feedback_score`
- Mean `feedback_response_rate`
- Mean `final_effectiveness`
- `effectiveness_tier`

This creates one row for each instructor. The resulting instructor-level features will be used as input variables for the machine learning model, while `effectiveness_tier` will be the target variable to predict.

```python
instructor_df = df.groupby("instructor_id")["final_effectiveness"].mean().reset_index()
```

```python
instructor_df 
```

```python
instructor_df["final_effectiveness"].describe()
```

```python
tier_lst=[]
for i in range(len(instructor_df)):
    eff=instructor_df.iloc[i]["final_effectiveness"]
    
    if eff < 0.505125:
        tier_lst.append("Low")
    elif eff <= 0.629490:
        tier_lst.append("Medium")
    else:
        tier_lst.append("High")

instructor_df["effectiveness_tier"]=tier_lst
instructor_df
```

### Effectiveness Tier Distribution

The 120 instructors were divided into three effectiveness tiers based on their final effectiveness scores.

- **Low:** 30 instructors
- **Medium:** 60 instructors
- **High:** 30 instructors

The resulting classes are reasonably balanced, with the Medium tier containing twice as many instructors as the Low and High tiers.

```python
instructor_df["effectiveness_tier"].value_counts()
```

```python
features = [
    "completion_rate",
    "avg_score_improvement",
    "avg_quiz_score",
    "not_drop_rate",
    "avg_watch_time",
    "assignment_submission_rate",
    "forum_activity_rate",
    "avg_feedback_score",
    "feedback_response_rate"
]
instructor_features_df = df.groupby("instructor_id")[features].mean().reset_index()
```

```python
instructor_features_df
```

```python
instructor_df = instructor_df.merge(
    instructor_features_df,
    on="instructor_id"
)

instructor_df
```

## Machine Learning Model

A **Random Forest Classifier** was used to predict the instructor effectiveness tier (`Low`, `Medium`, or `High`).

The instructor-level performance features were used as input variables, while `effectiveness_tier` was used as the target. `instructor_id` was excluded because it is only an identifier, and `final_effectiveness` was excluded because it was used to create the target tier.

The data was split into **80% training and 20% testing** using `random_state=42`.

### Model Performance

The model achieved:

- **Accuracy:** 100%
- **Precision:** 1.00 for all classes
- **Recall:** 1.00 for all classes
- **F1-score:** 1.00 for all classes

All **24 test instructors** were classified correctly.

### Why 100% Accuracy?

The perfect accuracy should be interpreted carefully. The effectiveness tiers were created from the same features that were given to the model.

The process was:

**Features → Effectiveness Score → Effectiveness Tier**

The model then learned:

**Features → Effectiveness Tier**

Therefore, the model is mainly learning to reproduce the scoring methodology we defined. The 100% accuracy shows that the model can reproduce these tiers effectively, but it does not independently validate that the effectiveness score is an objective measure of instructor performance.

### Feature Importance

The most important features were:

- `not_drop_rate`
- `completion_rate`
- `avg_score_improvement`
- `avg_quiz_score`

This is consistent with the scoring methodology because learner outcomes were given the highest weight.

Feature importance indicates how useful a feature was for prediction and does not imply causation.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
```

```python
x=df1.drop(columns=['instructor_id','final_effectiveness','effectiveness_tier'])
y=df1['effectiveness_tier']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
accu=accuracy_score(y_test,y_pred)
accu
```

```python
cl_r=classification_report(y_test, y_pred)
print(cl_r)
```

```python
feature_importance = pd.Series(
    model.feature_importances_,
    index=x.columns
).sort_values(ascending=False)

print(feature_importance)
```
